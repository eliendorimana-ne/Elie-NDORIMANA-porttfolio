# Iterative numerical solution of the optimization problem


$\min_{w \in R^{2}}\left\{F(w) = w_{1}w_{2}e^{-w_{1}^{2}-w_{2}^2}\right\}$

```python
import numpy as np
import numpy.linalg as la

import scipy.optimize as sopt

import matplotlib.pyplot as pt
from mpl_toolkits.mplot3d import axes3d


def myFun(w):
    

    return np.exp(-w[0]**2-w[1]**2)*(w[0]*w[1])

def gradFun(w):
    return np.array([np.exp(-w[0]**2-w[1]**2)*w[1]-2*np.exp(-w[0]**2-w[1]**2)*(w[0]**2)*w[1], \
                     np.exp(-w[0]**2-w[1]**2)*w[0]-2*np.exp(-w[0]**2-w[1]**2)*(w[1]**2)*w[0]])

w=np.array([1.0,2.0])
print(myFun(w))
print(gradFun(w))


''' fig = pt.figure()
ax = fig.gca(projection="3d")

xmesh, ymesh = np.mgrid[-2:2:50j,-2:2:50j]
fmesh = myFun(np.array([xmesh, ymesh]))
ax.plot_surface(xmesh, ymesh, fmesh)'''
```

## Q7(b) Determine the values of parameters W1, W2, W3, W4

```python
import numpy as np
import scipy.optimize as opt

x_data = np.array([0,1,2,3,4,5])
y_data = np.array([5.0,3.2,2.1,1.5,1.1,0.9])

def model(x, w1, w2, w3, w4):
    return w1*np.exp(-w3*x) + w2*np.exp(-w4*x)


def residuals(params, x, y):
    w1, w2, w3, w4 = params
    return y - model(x, w1, w2, w3, w4)


initial_guess = [1, 1, 0.5, 0.1]


result = opt.least_squares(residuals, initial_guess, args=(x_data, y_data))

w1, w2, w3, w4 = result.x


```

```python
print("Estimated parameters:")
print("w1 =", w1)
print("w2 =", w2)
print("w3 =", w3)
print("w4 =", w4)
```
